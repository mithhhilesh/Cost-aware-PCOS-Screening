# Corrected PCOS evaluation methodology

Review the evaluation settings before running. This is exploratory internal validation, not preregistration or clinical validation. Run from the repository directory with the evaluation dependencies installed. No historical results or dashboard artifacts are overwritten.

```python
from pathlib import Path
import subprocess, sys, json
import pandas as pd
assert Path("evaluate.py").exists(), "Open this notebook from the repository folder."
```

## Declare settings before viewing results
90% is an illustrative inner sensitivity target, not a clinical recommendation. Record all deviations. A repeated-patient dataset requires a verified grouping column.

```python
DATA = Path("PCOS_dataset.csv").resolve()
OUTPUT = Path("evaluation_runs/run_01")
TARGET_SENSITIVITY = 0.90
GROUP_COLUMN = None  # Set only after verifying the patient identifier.
INCLUDE_RF = False  # Declare candidate set before running.
assert DATA.exists(), "Set DATA to the original CSV."
assert not OUTPUT.exists(), "Choose a new run name; do not overwrite an old evaluation."
```

```python
command = [sys.executable, "evaluate.py", "--data", str(DATA),
           "--output", str(OUTPUT), "--target-sensitivity", str(TARGET_SENSITIVITY)]
if GROUP_COLUMN:
    command += ["--group-column", GROUP_COLUMN]
if INCLUDE_RF:
    command += ["--include-rf"]
subprocess.run(command, check=True)
```

```python
print((OUTPUT / "REPORT.md").read_text(encoding="utf-8"))
display(pd.read_csv(OUTPUT / "metrics.csv"))
display(pd.read_csv(OUTPUT / "paired_differences.csv"))
```
